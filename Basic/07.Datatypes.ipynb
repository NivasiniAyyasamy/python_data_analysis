**String Formatting**

### format

```python
role = 'Data Analyst'
skill = 'Python'
```

```python
# format(...)
#  |      S.format(*args, **kwargs) -> str
#  |
#  |      Return a formatted version of S, using substitutions from args and kwargs.
#  |      The substitutions are identified by braces ('{' and '}').

'Role: {}'.format(role)
```

```python
#Create a string: 'Role: Data Analyst; Skill Required: Python'
'Role: {}; Skill Required: {}'.format(role, skill)
#'Role: {role}; Skill Required: {skill}'.format(role = role, skill = skill)
```

#### f-String - Formatted string Literals

```python
f'Role: {role}; Skill Required: {skill}'
```

#### %-formatting (printf-style string formatting)

```python
'Role: %s; Skill Required: %s' % (role, skill) # Not recommended
#'Role: %s' % (role)
```

###Slicing

### Unpacking
Assign each value in an iterable to a variable in a single statement.

```python
print(job_skills)
skill1, skill2, skill3 = job_skills
print(skill1)
print(skill2)
print(skill3)
```

```python
skill_concerned, *skill_dont_care = job_skills #Assign the first value to the the first variable and * is the unpacking symbol that assigns all the other items as a list to the second variable
```

```python
print(skill_concerned)
print(skill_dont_care)
```

## Dictionary
* Usage: For key-value pairs
* Syntax: {'key1' : 'value1', 'key2' : 'value2', ...}
* Type of Data: Mapping
* Order: Unordered
* Indexing: Yes(by key)
* Duplicate values: Duplicate values are allowed but keys cannot
* Mutability: Mutable


```python
#{'database': 'PostgreSQL'}
#{'database': ['PostgreSQL', 'SQL]}
#{'database': [['PostgreSQL', 'SQL'], 4, 'string', (1,2)]}
job_type_skills = {
    'database': "SQL",
    'language': 'Python',
    'library': 'Pandas'
}
job_type_skills
```

```python
job_type_skills['database']
```

```python
print(job_type_skills.keys())
print(job_type_skills.values())
print(job_type_skills.items())
```

```python
job_type_skills.pop('library')
job_type_skills
```

```python
job_type_skills.update({'cloud': 'AWS'})
job_type_skills #This and the above step proves dict datatype is mutable
```

```python
#Instaed Update we can use the below step to add a key - value pairs
job_type_skills['version_control'] = 'git'
job_type_skills
```

## Sets

1. Sets are very similar to list. The Exception here is instead of using a square brackets, we use curly brackets
2. Sets are mutable
3. However, sets are unique. That is, it will not allow duplicate values in it.

* Usage: For unique items
* Syntax: {item1, item2,....}
* Type of data: Set
* Order: Unordered
* Indexing: No
* Duplicate values: Not allowed
* Mutability: Yes, mutable

```python
job_skills = { 'tableau', 'sql', 'python', 'sql', 'statistics'} # Any datatype can be added
job_skills # duplicates will be removed and the order might not be the same when we created list
```

```python
job_skills.add('looker') #suppose if we provide a value that is already present in the set. The code runs but does not the item.
job_skills
```

```python
job_skills.pop() #Unlike list, here we cannot specify the element to be removed.
#Instead pop randomly removes an elemnt from the set. set() - new empty set object
#set(iterable) - new set object
```

```python
# If we have to remove a specific element then prodive 'remove'
job_skills.remove('looker')
```

```python
job_skills
```

```python
# Suppose if we already have a list that is with duplicates, then
skills_list = ['Python', 'sql', 'tableau', 'sql', 'statistics', 'tableau' ]
set(skills_list)
```

```python
# Again if we have to chane the set of unique values to a list
list(set(skills_list))
```

## Tuples

1. Tuples are inside (paranthesis), any datatype can be added to the tuple.
2. Unlike list, we cannot add or remove an item from tuple

* Usage: For fixed data
* Syntax: (item1, item2, item3, ...)
* Type of Data: Sequence
* Order: Ordered
* Indexing: Yes, by index
* Duplicate values: Allowed
* Mutability: Unmutable

```python
nivas_skills = ("Python", "sql", "PowerBI")
```

```python
# Slicing works same as list
nivas_skills[0:2]
```

```python
nivas_skills = ("Python", "sql", "PowerBI")
gk_skills = ("Marketing", "Excel",)
all_skills = nivas_skills + gk_skills # Can add new tuple to the existing one by nivas_skills += gk_skills
all_skills
```

```python
nivas_skills = ("Python", "sql", "PowerBI")
print(id(nivas_skills))
gk_skills = ("Marketing", "Excel",)
nivas_skills += gk_skills
nivas_skills # obviously now the value for nivas_skills be changed, to identify the changes we can use id function
print(id(nivas_skills))
```

```python
# But for list the same item is changed
nivas_skills = ["Python", "sql", "PowerBI"]
print(id(nivas_skills))
gk_skills = ["Marketing", "Excel",]
nivas_skills += gk_skills
print(id(nivas_skills))
```
