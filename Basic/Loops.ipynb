## FOR

For loop iterates through a iterablesuch as list, tuple, set, dict strings etc

```python
#for i in iterable:
  #statement

#Iterable here can be of any datatype like dict, list, tuple, set, string
```

```python
numbers = [1, 2, 3, 4, 5]
for number in numbers:
  if number >= 3:
    print(number)
```

```python
years_experience = {
    'Nivasini': 4,
    'Dharani': 3,
    'Archana': 5,
    'Gokul': 6
}
```

```python
for person in years_experience:
  print(person)
```

```python
for value in years_experience.values():
  print(value)
```

```python
for key, value in years_experience.items():
  print(key, value) # returns a set like object
```

```python
#Applicant is having greater than 5 years of experience
for key, value in years_experience.items():
  if value >= 5:
    print(key)
```

```python
for _ in range(10):
  print(_)
```

## WHILE

While loop checks for the condition, if true then the code inside it is processed.


```python
years = 0
index = 0

years_list = list(years_experience.items())
print(years_list)

while years <= 6:
  key, years = years_list[index]
  if years >= 5:
    print(key, years)
  index += 1
  years += 1
```
