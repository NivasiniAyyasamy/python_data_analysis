# List Comprehension

Easily creating a list of numbers

```python
# Create a list from 0 - 10
numbers = [x for x in range(10)]
```

```python
numbers
```

```python
# Suppose if we didn't know list comprehension to create a list

numbers =[]
for x in range(10):
  numbers.append(x)
numbers # This is unnecessary steps
```

```python
numbers = [float(x) for x in range(10)]
numbers
```
